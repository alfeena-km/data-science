```python
rows = 5

for i in range(1, rows + 1):
    for j in range(i):
        print("* ", end="")
    print()
```

```python
rows = 5

for i in range(rows, 0, -1):
    for j in range(i):
        print("* ", end="")
    print()
```

```python
rows = 5

# Upper half of the diamond
for i in range(1, rows + 1):
    print(" " * (rows - i), end="")  # Print leading spaces
    print("* " * i)  # Print stars

# Lower half of the diamond (inverted, excluding the middle row)
for i in range(rows - 1, 0, -1):
    print(" " * (rows - i), end="")  # Print leading spaces
    print("* " * i)  # Print stars
```
